1. В этом блоке подключаем необходимые библиотеки: openpyxl для работы с Excel и matplotlib.pyplot для построения графиков.

```python
# Тихонович Данила Вариант 5

from openpyxl import load_workbook, Workbook
import matplotlib.pyplot as plt
```

2. Загружаем Excel-файл data/students.xlsx, считываем заголовки и все строки данных. Преобразуем каждую строку в словарь для удобства доступа по названию предмета

```python
# Тихонович Данила Вариант 5
wb = load_workbook("../data/students.xlsx")
ws = wb.active

headers = [cell.value for cell in ws[1]]

records = []
for row in ws.iter_rows(min_row=2, values_only=True):
    records.append(dict(zip(headers, row)))
```

Вариант 5 

1. Вычислить средний балл по каждому предмету: Математика, Физика, Информатика.

```python
# Тихонович Данила Вариант 5

subjects = ['Математика', 'Физика', 'Информатика']
average = {}

for subject in subjects:
    total = 0
    count = 0
    for record in records:
        if record[subject] is not None:
            total += record[subject]
            count += 1
    average[subject] = total / count        

print ("Средний балл по предметам: ")
for subject, avg in average.items():
    print (f"{subject}: {avg}")
```

2. Построить столбчатую диаграмму, где по оси X — предметы, по оси Y — средний балл. 

```python
# Тихонович Данила Вариант 5
plt.figure(figsize=(8,5))
plt.bar(average.keys(), average.values(), color=['skyblue', 'salmon', 'lightgreen'])
plt.ylim(0,10)  
plt.show()
```

3. 3. Подписать столбцы и добавить сетку.

```python
# Тихонович Данила Вариант 5
plt.figure(figsize=(8,5))
plt.bar(average.keys(), average.values(), color=['skyblue', 'salmon', 'lightgreen'])
plt.xlabel('Предметы')
plt.ylabel('Средний балл')
plt.title('Средний балл по предметам')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(0,10)  
plt.show()
```

Задание 4.
 
Работа с данными

```python
# Тихонович Данила Вариант 5
result_wb = Workbook()
result_ws = result_wb.active
result_ws.title = "Средние баллы поо предметам"

result_ws.append(['Предмет', 'Средний балл'])

for subject, avg in average.items():
    result_ws.append([subject, round(avg, 2)])

result_wb.save("../data/students_result.xlsx")
print("Средние баллы были сохранены в students_result.xlsx")
```
